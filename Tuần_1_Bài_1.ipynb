```python
# ==============================================================================
# BÀI 1: THỰC HÀNH TẢI TỆP BREAST CANCER TRONG SCIKIT-LEARN VÀ XỬ LÝ VỚI PANDAS
# ==============================================================================

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.datasets import load_breast_cancer

# --- BƯỚC 0: TẢI DỮ LIỆU TỪ SCIKIT-LEARN VÀO PANDAS DATAFRAME ---
cancer = load_breast_cancer()
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)
# 0: malignant (ung thư/ác tính), 1: benign (lành tính)
df['target'] = cancer.target

# --- YÊU CẦU 1: XEM DANH SÁCH CÁC THUỘC TÍNH CỦA DỮ LIỆU ---
print("=" * 60)
print("1. DANH SÁCH CÁC THUỘC TÍNH:")
print(list(df.columns))

# --- YÊU CẦU 2: XEM KÍCH THƯỚC DỮ LIỆU VÀ NHÃN ---
print("\n" + "=" * 60)
print("2. KÍCH THƯỚC DỮ LIỆU VÀ NHÃN:")
print(f"- Kích thước toàn bộ bảng (DataFrame): {df.shape[0]} dòng, {df.shape[1]} cột")
print(f"- Kích thước ma trận đặc trưng X: {cancer.data.shape}")
print(f"- Kích thước vector nhãn y: {cancer.target.shape}")
print(f"- Tên các nhãn phân loại: {cancer.target_names.tolist()} (0: Ung thư, 1: Lành tính)")

# --- YÊU CẦU 3: LỌC RA NHỮNG MẪU CÓ BỆNH UNG THƯ (MALIGNANT - 0) ---
print("\n" + "=" * 60)
df_ung_thu = df[df['target'] == 0]
print(f"3. SỐ LƯỢNG MẪU CÓ BỆNH UNG THƯ: {len(df_ung_thu)} mẫu")
print("- 3 mẫu đầu tiên có bệnh ung thư:")
display(df_ung_thu.head(3))

# --- YÊU CẦU 4: LỌC RA NHỮNG MẪU LÀNH TÍNH (BENIGN - 1) ---
print("\n" + "=" * 60)
df_lanh_tinh = df[df['target'] == 1]
print(f"4. SỐ LƯỢNG MẪU LÀNH TÍNH: {len(df_lanh_tinh)} mẫu")
print("- 3 mẫu đầu tiên lành tính:")
display(df_lanh_tinh.head(3))

# --- YÊU CẦU 5: TÍNH TOÁN ĐỘ TƯƠNG QUAN CỦA TỪNG THUỘC TÍNH VỚI NHÃN BỆNH ---
print("\n" + "=" * 60)
print("5. ĐỘ TƯƠNG QUAN CỦA TỪNG THUỘC TÍNH VỚI THUỘC TÍNH BỆNH (TARGET):")
corr_with_target = df.corr()['target'].drop('target').sort_values(ascending=False)
print(corr_with_target)

# --- YÊU CẦU 6: VẼ BIỂU ĐỒ THỂ HIỆN ĐỘ TƯƠNG QUAN GIỮA CÁC THUỘC TÍNH ---
print("\n" + "=" * 60)
print("6. ĐANG VẼ BIỂU ĐỒ TƯƠNG QUAN (HEATMAP)...")
plt.figure(figsize=(16, 12))
sns.heatmap(df.corr(), cmap='coolwarm', annot=False, cbar=True)
plt.title("Biểu đồ nhiệt (Heatmap) thể hiện độ tương quan giữa các thuộc tính", fontsize=15, pad=15)
plt.tight_layout()
plt.show()

# --- YÊU CẦU 7: TRỰC QUAN HÓA TỶ LỆ NHÃN ---
print("\n" + "=" * 60)
print("7. ĐANG TRỰC QUAN HÓA TỶ LỆ NHÃN...")
label_names = ['Có bệnh ung thư\n(Malignant)', 'Lành tính\n(Benign)']
counts = [len(df_ung_thu), len(df_lanh_tinh)]
colors = ['#e74c3c', '#2ecc71']

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

# Biểu đồ cột (Bar chart)
bars = axes[0].bar(label_names, counts, color=colors, width=0.5, edgecolor='black')
axes[0].set_title("Số lượng mẫu theo từng nhãn", fontsize=13)
axes[0].set_ylabel("Số lượng mẫu")
for bar in bars:
    yval = bar.get_height()
    axes[0].text(bar.get_x() + bar.get_width()/2.0, yval + 5, int(yval), ha='center', va='bottom', fontweight='bold')

# Biểu đồ tròn (Pie chart)
axes[1].pie(counts, labels=label_names, autopct='%1.1f%%', startangle=90,
            colors=colors, explode=(0.05, 0), shadow=True)
axes[1].set_title("Tỷ lệ phần trăm giữa hai nhãn", fontsize=13)

plt.suptitle("Trực quan hóa tỷ lệ nhãn (Breast Cancer Dataset)", fontsize=15, y=1.03)
plt.tight_layout()
plt.show()
print("=" * 60)
```
