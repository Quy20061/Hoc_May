### 1. Đọc dữ liệu, hiển thị thuộc tính và 10 dòng đầu tiên

```python
import pandas as pd

# Tải dữ liệu
file_path = '/content/drive/MyDrive/breast-cancer.csv'
df = pd.read_csv(file_path)

# Hiển thị các thuộc tính (cột) trong tệp
print("Các thuộc tính trong tệp dữ liệu:")
print(df.columns.tolist())
print("\nThông tin tổng quan:")
df.info()

# Hiển thị 10 dòng đầu tiên
print("\n10 dòng đầu tiên:")
display(df.head(10))
```

### 2. Xóa các dòng có dữ liệu thiếu

```python
# Kiểm tra số lượng giá trị thiếu trước khi xóa
print("Số lượng giá trị thiếu trước khi xóa:")
print(df.isnull().sum())

# Xóa các dòng chứa giá trị thiếu
df_cleaned = df.dropna()

# Xác nhận lại kích thước tập dữ liệu
print(f"\nKích thước tập dữ liệu ban đầu: {df.shape}")
print(f"Kích thước tập dữ liệu sau khi xóa dòng thiếu: {df_cleaned.shape}")
```

### 3. Chia dữ liệu thành tập huấn luyện và tập kiểm tra

```python
from sklearn.model_selection import train_test_split

# Chia tập dữ liệu (70% train, 30% test làm ví dụ)
train_df, test_df = train_test_split(df_cleaned, test_size=0.3, random_state=42)

print("20 dòng đầu tiên của tập huấn luyện (Train Set):")
display(train_df.head(20))

print("\n20 dòng đầu tiên của tập kiểm tra (Test Set):")
display(test_df.head(20))
```

### 4. Chuẩn hóa dữ liệu (Standardization) và Mã hóa một nóng (One-hot encoding)

```python
from sklearn.preprocessing import StandardScaler
import numpy as np

# Xác định các cột số (numerical) và các cột phân loại (categorical)
# Giả định cột đầu tiên (hoặc cột ID/Diagnosis) là phân loại, chúng ta sẽ tự động phân loại cột.
numerical_cols = df_cleaned.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df_cleaned.select_dtypes(exclude=[np.number]).columns.tolist()

print(f"Cột số: {numerical_cols}")
print(f"Cột phân loại: {categorical_cols}")

# Sao chép dữ liệu để xử lý
df_processed = df_cleaned.copy()

# 1. Chuẩn hóa dữ liệu (Standardization)
scaler = StandardScaler()
if numerical_cols:
    df_processed[numerical_cols] = scaler.fit_transform(df_cleaned[numerical_cols])

# 2. Mã hóa biến phân loại (One-hot encoding)
if categorical_cols:
    df_processed = pd.get_dummies(df_processed, columns=categorical_cols, drop_first=True)

# Hiển thị lại 10 dòng dữ liệu đầu tiên sau khi xử lý
print("\n10 dòng đầu tiên sau khi chuẩn hóa và mã hóa:")
display(df_processed.head(10))
```

### 5. Tính toán thống kê cơ bản

```python
# Thống kê cơ bản (mean, median, std) cho các thuộc tính số ban đầu
stats_df = df_cleaned[numerical_cols].describe().loc[['mean', '50%', 'std']]
stats_df.rename(index={'50%': 'median'}, inplace=True)

print("Các thông số thống kê cơ bản của dữ liệu gốc:")
display(stats_df)
```

### 6. Vẽ biểu đồ phân phối

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Chọn tối đa 3 thuộc tính số đầu tiên để vẽ biểu đồ phân phối
cols_to_plot = numerical_cols[:3]

plt.figure(figsize=(15, 5))
for i, col in enumerate(cols_to_plot, 1):
    plt.subplot(1, len(cols_to_plot), i)
    sns.histplot(df_cleaned[col], kde=True, color='skyblue')
    plt.title(f'Phân phối của {col}')
    plt.xlabel(col)
    plt.ylabel('Tần suất')

plt.tight_layout()
plt.show()
```
