```python
import pandas as pd
import numpy as np

# 1. Đọc dữ liệu
df = pd.read_csv('/content/drive/MyDrive/Học Máy (1)/DL bài 2/sample_data.csv')

# 2. Khám phá dữ liệu ban đầu
print("--- THÔNG TIN TỔNG QUAN ---")
df.info()
display(df.describe(include="all"))

# 3. Kiểm tra tỷ lệ giá trị thiếu (%)
print("\n--- TỶ LỆ GIÁ TRỊ THIẾU (%) ---")
missing_ratio = df.isnull().mean() * 100
print(missing_ratio)

# 4. Xử lý giá trị thiếu (Thứ tự đúng: Điền trước, sau đó mới xóa)
# Điền giá trị thiếu cho cột 'Age' bằng giá trị trung bình (Mean)
if 'Age' in df.columns:
    df['Age'] = df['Age'].fillna(df['Age'].mean())

# Điền giá trị thiếu cho cột 'Gender' bằng giá trị xuất hiện nhiều nhất (Mode)
if 'Gender' in df.columns:
    df['Gender'] = df['Gender'].fillna(df['Gender'].mode()[0])

# Xóa các dòng còn sót lại giá trị thiếu ở các cột quan trọng khác (nếu có)
df.dropna(how='any', inplace=True)

# 5. Xử lý dữ liệu trùng lặp
print(f"\nSố lượng dòng trùng lặp trước khi xóa: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)

# 6. Chuẩn hóa định dạng chuỗi (Gender)
if 'Gender' in df.columns:
    # Xóa khoảng trắng thừa và chuyển toàn bộ về chữ thường
    df['Gender'] = df['Gender'].astype(str).str.strip().str.lower()

# 7. Chuẩn hóa định dạng thời gian (Date)
if 'Date' in df.columns:
    # errors='coerce' giúp chuyển các dữ liệu sai định dạng thành NaT thay vì báo lỗi crash code
    df['Date'] = pd.to_datetime(df['Date'], errors='coerce')

# 8. Lưu file sau khi làm sạch
df.to_csv('sample_data_cleaned.csv', index=False)
print("\n=> Đã làm sạch dữ liệu thành công và lưu vào 'sample_data_cleaned.csv'!")
df = pd.read_csv('/content/sample_data_cleaned.csv')
df.head()
```
